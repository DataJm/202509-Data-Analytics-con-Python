## Cómo seleccionar los  NA ?

```python
titanic_missing_age = titanic[titanic["age"].isna()]
titanic_missing_age
```

### Exportar a CSV

# Cómo filtrar (drop) las filas con NA ? 

```python
# El caracter de "~" invierte el sentido de una serie booleana ( lógica )
~titanic["age"].isna()
```

```python
titanic_valid_age = titanic[~titanic["age"].isna()]
titanic_valid_age
```

```python
titanic_valid_age.info()
```

# Seleccionar toda la data válida

En ciencia de datos, en la parte de modelado de datos. Para eso existe un método que "tira" todos los datos que presenten algún valor ausente. En otras palabras, retiene sólo "datos validos"

```python
titanic_valid_values = titanic.dropna()

titanic_valid_values
```

```python
titanic.dropna(axis=1)
```

# Imputar valores ausentes

**Nota**: Imputar significa estimar, aqui debemos recordar que vamos a introducir un margen de error derivado de nuestra imputación. 

```python
# Sobre escribe los valores NA
titanic["age_imputed"] = titanic["age"].fillna(round(titanic["age"].mean(),0))
```
