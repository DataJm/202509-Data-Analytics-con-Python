## 🕸️ Web Scraping: Conocimientos y Recomendaciones

El **web scraping** permite automatizar la extracción de información de sitios web.  
Para hacerlo correctamente —sin errores ni bloqueos— es necesario dominar varias áreas técnicas y éticas.

---

### 🧠 Conocimientos Clave

| Área | Qué debes saber | Por qué importa |
|------|------------------|-----------------|
| **Python** | Estructuras de datos, manejo de errores, requests, loops | Base para crear scrapers confiables |
| **HTTP y redes** | Métodos GET/POST, headers, cookies, sesiones | Para simular peticiones reales del navegador |
| **HTML / DOM / Selectores** | Etiquetas, clases, `CSS selectors`, `XPath` | Para localizar y extraer elementos |
| **JavaScript y APIs** | Entender sitios dinámicos y llamadas AJAX | Muchas webs cargan datos por JS, no en HTML |
| **Limpieza y almacenamiento** | Uso de Pandas, CSV, JSON, bases de datos | Para transformar y guardar los datos extraídos |
| **Buenas prácticas** | `robots.txt`, pausas, ética y legalidad | Evita bloqueos y problemas legales |

---

### 📘 Libros Recomendados

1. **_Web Scraping with Python (3rd Ed.)_ — Ryan Mitchell**  
   Completo, moderno, cubre scraping con Selenium, APIs y aspectos legales.

2. **_Python Web Scraping Cookbook_ — Michael Heydt**  
   Libro tipo “recetario” con soluciones prácticas a problemas comunes.

3. **_Python Web Scraping_ — Katharine Jarmul & Richard Lawson**  
   Ideal para principiantes que quieran avanzar a nivel profesional.

---

### 🧩 Consejos Finales

- Empieza con **sitios estáticos** y `requests + BeautifulSoup`.  
- Usa **Selenium o Playwright** solo cuando sea necesario (sitios dinámicos).  
- Sé paciente y **respetuoso** con los servidores: espera entre peticiones.  
- Guarda logs y errores, los scrapers fallan más de lo que parecen.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
```

### No esperar a que el sitio web carge con completo


```python
import time
```

```python
options = webdriver.ChromeOptions()
options.page_load_strategy = "none"

driver = webdriver.Chrome(options=options)
driver.get("https://finance.yahoo.com/quote/ES%3DF/history/")

time.sleep(10)

tables = pd.read_html(driver.page_source)
driver.quit()
```
