### Yahoo Finance parte 2

- Pareciera que yahoo finance tiene una protección en la que deja esperando indefinidamente a los navegadores web controlados por python.
- Podemos controlar que la conexión se detenga después de un tiempo de carga

```python
import time

options = webdriver.ChromeOptions()
options.page_load_strategy = "none"

driver = webdriver.Chrome(options=options)
driver.get("https://finance.yahoo.com/quote/ES%3DF/history/")

time.sleep(10)

tables = pd.read_html(driver.page_source)
driver.quit()
```
