# Loc & iLoc

```python
tienda["Nombre"] # A los [] les puedo pasar un nombre de columna
```

```python
tienda[tienda["Ventas"]>=100] # A los [] les puedo pasar un vector lógico (un filtro)
```

### En los data frames, los [ ] son ambigüos

```python
tienda
```

### Loc

Busqueda por "nombre"

```python
# tienda.loc[ filas, columnas ]

# tienda.loc[0,"Nombre"]

tienda.loc[0,:]
```

```python
# tienda.loc[:,0] # Esto produce error porque necesita el nombre de la columna
```

### iLoc

Busqueda por posición

```python
tienda.iloc[:,1]
```

```python
tienda[tienda["Ventas"]>=100].iloc[:,0]
```

```python
# tienda[tienda.iloc[:,2]>=100].iloc[:,0]

filtro_ventas_mayores_100 = tienda.iloc[:,2]>=100

tiendas_mayores_a_100 = tienda[filtro_ventas_mayores_100]

nombre_tiendas_mayores_a_100 = tiendas_mayores_a_100.iloc[:,0]
```

```python
nombre_tiendas_mayores_a_100
```

Al hacer test de "igual a..." no cometer el error de "asignar"

Este código está sobreescribiendo a todos los registros / filas del dataset para ser iguales a "male"

## Método .value_counts()

```python
# Filtro por todos los registros donde sex = male, selecciono la columna "embarked" ,
# finalmente uso el mètodo value_counts()

titanic[titanic["sex"] == "male"]["embarked"].value_counts()
```
